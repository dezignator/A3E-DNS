# 02 — AEE / SE(3)ⁿ Analysis

**A³E-DNS Thesis Artifact · Chapter 6 · §6.4, Eqs. 119–122**

This notebook exercises the Adaptive Evolutionary Encryption engine
and demonstrates the SE(3)ⁿ state evolution. It does **not** claim
security: the schedule is a heuristic construction (§6.4.4); the
formal security argument is reserved for the KEM layer
(Proposition 1, §5.2.6).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import numpy as np
from src.models.aee_engine import AEEEngine, AEEEngineConfig

engine = AEEEngine(AEEEngineConfig(se3_dim=2))
state = engine.initialize()
print('initial R shape:', state.R.shape)
print('initial step   :', state.step)
print('initial key len:', len(state.session_key), 'bytes')
```

```python
features = np.zeros(64)
for i, kl in enumerate([0.0, 0.1, 0.5, 1.0, 2.0]):
    state = engine.step(features, kl_divergence=kl)
    print(f"step {state.step:2d}  kl={kl:.2f}  sigma={engine._sigma(kl):.6f}")
```

### σ(t) is strictly positive

Eq. 122 and §6.4.4 require σ(t) to be bounded away from zero. This is a design choice, not a security claim.

```python
for kl in [0.0, 0.01, 0.1, 1.0, 10.0, 1000.0]:
    s = engine._sigma(kl)
    assert s > 0.0, f'sigma({kl}) must be > 0, got {s}'
    print(f'kl={kl:8.2f}  sigma={s:.8f}')
```
